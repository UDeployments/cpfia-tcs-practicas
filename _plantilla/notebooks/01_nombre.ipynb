## 2. Configuración

Todos los imports y rutas en una sola celda. Las rutas son relativas a la carpeta de la práctica.

```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

BASE_DIR = Path.cwd().parent
RAW_DIR = BASE_DIR / "data" / "raw"
PROCESSED_DIR = BASE_DIR / "data" / "processed"
OUTPUT_DIR = BASE_DIR / "outputs"
for d in (RAW_DIR, PROCESSED_DIR, OUTPUT_DIR):
    d.mkdir(parents=True, exist_ok=True)

pd.set_option("display.max_columns", 50)
```

## 3. Carga de datos

**Fuente:** nombre del dataset y enlace u origen.

**Descripción:** qué representa cada fila y las columnas principales.

Reemplaza el ejemplo por `pd.read_csv(RAW_DIR / "archivo.csv")` o `pd.read_excel(...)`.

```python
rng = np.random.default_rng(42)
df = pd.DataFrame({
    "categoria": rng.choice(["A", "B", "C"], size=100),
    "valor": rng.normal(50, 10, size=100).round(2),
})
df.head()
```

## 4. Exploración de datos (EDA)

Forma, tipos, nulos, estadísticos y primeras gráficas.

```python
print(df.shape)
df.info()
df.describe()
```

## 5. Limpieza y transformación

Explica cada decisión: qué se elimina, imputa o transforma y por qué.

```python
df_clean = df.dropna().copy()
df_clean.to_csv(PROCESSED_DIR / "datos_limpios.csv", index=False)
```

## 6. Análisis / Modelado

Describe la técnica usada y por qué.

```python
resumen = df_clean.groupby("categoria")["valor"].agg(["count", "mean", "std"]).round(2)
resumen
```

## 7. Resultados

Gráficas y tablas finales. Exporta lo relevante a `outputs/`.

```python
fig, ax = plt.subplots(figsize=(6, 4))
resumen["mean"].plot(kind="bar", ax=ax)
ax.set_title("Valor promedio por categoría")
ax.set_ylabel("Valor")
fig.savefig(OUTPUT_DIR / "promedio_por_categoria.png", dpi=150, bbox_inches="tight")
plt.show()
```
